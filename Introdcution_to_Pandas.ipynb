**Introduction to Pandas for Data Analysis**

Pandas is a powerful library for data manipulation and analysis in Python. This tutorial covers essential operations related to appending data, merging, and joining DataFrames.

**pd.concat()**

Used to concatenate (combine) multiple DataFrames along rows or columns.

**pd.Series()**

Used to create a one-dimensional labeled array (like a column in a DataFrame).

```python
import pandas as pd
```

1. Appending a List of Dictionaries or Series to an Existing DataFrame

```python
import pandas as pd

# Existing DataFrame
df = pd.DataFrame({'ID': [1, 2], 'Name': ['Alice', 'Bob']})

# List of dictionaries to append
new_data = [{'ID': 3, 'Name': 'Charlie'}, {'ID': 4, 'Name': 'David'}]

#pd.concat(#tableName , #NewRow)

# Append and reset index
df = pd.concat([df, pd.DataFrame(new_data)], ignore_index=True)
print(df)
```

```python
# Existing DataFrame
df = pd.DataFrame({'ID': [1, 2, 3, 4], 'Name': ['Alice', 'Bob', 'Charlie', 'David']})

# Creating a new Series
new_row = pd.Series({'ID': 5, 'Name': 'Eve'})

# Convert Series to DataFrame and use pd.concat()
df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)

print(df)
```

**pd.merge() in Pandas**

**pd.merge()** is used to combine two DataFrames based on a common column (like SQL joins).

**1. Outer Join (how='outer')**

Keeps all records from both DataFrames, filling missing values with NaN where there's no match.

**2. Left Join (how='left')**

Keeps all records from the left DataFrame and only matching records from the right.

**3. Right Join (how='right')**

Keeps all records from the right DataFrame and only matching records from the left.

**2. Joining Two DataFrames Along Rows and Merging with Another DataFrame**

```python
# Creating two DataFrames
df1 = pd.DataFrame({'ID': [1, 2], 'Name': ['Alice', 'Bob']})
df2 = pd.DataFrame({'ID': [3, 4], 'Name': ['Charlie', 'David']})

# Concatenating along rows
df_combined = pd.concat([df1, df2], axis=0, ignore_index=True)
print(df_combined)
```

**3. Joining Two DataFrames Using the Common Column**

```python
# Merging with another DataFrame
df3 = pd.DataFrame({'ID': [1, 2, 3, 4], 'Age': [25, 30, 22, 28]})
df_merged = df_combined.merge(df3, on='ID')
print(df_merged)
```

```python
# Creating DataFrames with a common column 'ID'
df1 = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
df2 = pd.DataFrame({'ID': [2, 3, 4], 'Age': [30, 22, 28]})

# Performing an inner join
df_joined = df1.merge(df2, on='ID')
print(df_joined)
```

**4. Joining DataFrames with Matching Records from Both Sides (Outer Join)**

```python
# Performing an outer join
df_outer = df1.merge(df2, on='ID', how='outer')
print(df_outer)
```

**5. Left Join Using Keys from Left DataFrame Only**

```python
# Performing a left join
df_left = df1.merge(df2, on='ID', how='left')
print(df_left)
```

**6. Right Join Using Keys from Right DataFrame Only**

```python
# Creating DataFrames with a common column 'ID'
df1 = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
df2 = pd.DataFrame({'ID': [2, 3, 4], 'Age': [30, 22, 28]})

# Performing a right join
df_right = df1.merge(df2, on='ID', how='right')
print(df_right)
```
