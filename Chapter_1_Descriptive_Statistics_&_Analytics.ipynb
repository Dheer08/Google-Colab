# What is the variable?
A variable is any quantity which can be measured or counted.

#Different types of variables



image.png


1.  **A nominal variable** is a kind of variable to name the sample i.e. name of customers, items, products,etc.
2.    **An ordinal variable** is a categorical variable where the poissble outcomes can be sorted.

3. **A discrete variable** is a variable in which possible outcomes are countable.

4. **A continuous variable** is a variable in which possible values are uncountable.

## Quantitative variable:
A quantitative variable is a variable which takes the numercial values.

Quantitative variables are divided into two types: **discrete** and **continuous**. The difference is explained as follows:

###Discrete variables:
A quantitative discrete variable is  a variable  which is finitely countable. Here are some examples of discrete variables:

* Number of passangers who use a particular airline
* Number of typos created by a typing machine.
* Number of sensors in a network

Remember that the number of passangers never be 3.5, it is always finite number and thus countable.



### Continuous variable:
On the other hand, quantitative continuous variable is a  variable for which the values are not countable and have an infinite number of possibilities. For example:

* Age
* Weight
* Height

For simplicity, we use years for age. However, a 25-year-old man could actually be 25 years, 2 months, 11 days, 2 hours, 5 minutes, 9 seconds, 5 milliseconds old.
The fact that the possible values of a continuous variable is uncountable.

## Qualitative variable:
In opposition to quantitative variables, qualitative variables (also referred as categorical variables or factors in R) are variables that are not numerical.

Qualitative variables are divided into two types: **nominal** and **ordinal**.



### Nominal variable:
A qualitative nominal variable is a non-numerical variable which cannot be ordered or labled. For example, **the name of customers is nominal**.  


### Ordinal variable:

A qualitative ordinal variable is a non-numerical variable which can be ordered. For instance, the **temperature** can be implied in three level high, medium, low.

###Interquartile Range (IQR)
The IQR is the range of 50% of middle data points when are sorted from minmum to maximum. To find the interquartile range (IQR), ​first find the median (middle value) of the lower and upper half of the data. These values are quartile 1 (Q1) and quartile 3 (Q3). The IQR is the difference between Q3 and Q1.

image.png

**Illustrative Example 8:** Let D be D={1,4,3,2,8,9}, compute IQR.

D={1,4,3,2,8,9}={1,2,3,4,8,9}

Q1=2,

Q3=8,

IQR=Q3-Q1=8-2=6

##outlier detection

In data analysis, outlier detection is the task of identification of rare data points which are significanlty different from the majority of the data.

Techniques of outlier detection:
1. Outlier detection using Chebechev's rule using mean and std
2. Outlier detection using boxplot



image.png

###Outliers detection using Z-scores:
We can use one of the following intervals to specify outliers:

1. one-sigma standard interval: (mean-sd,mean+sd)
2. two-sigma standard interval: ((mean-2*sd,mean+2*sd)
3. three-sigma standard interval: ((mean-3*sd,mean+3*sd)


```python
%load_ext rpy2.ipython
```

**Illustrative Example 15:**
Specify the outliers of **disp** in **mtcars** dataset if any, using one-sigma standard interval.

```python
%%R
X=mtcars$disp
#(mean-sd,mu+sd) one-sigma interval estimation

Lbound=mean(X)-sd(X); Ubound=mean(X)+sd(X) # one-sigma interval estimation
print(c(Lbound,Ubound))

Out=c() # empty array
for (j in X){
   if(j<Lbound|j>Ubound){o=j}else{o=NULL}
   Out=c(Out,o) # append o to the vector of Out
}
Out # an array of outliers

```

```python
%%R
data=read.csv('/content/boston_house_prices.csv')
head(data)
X=data$TAX
dim(data)
length(X)
```

```python
%%R
# two sigma interval estimation
L=mean(X)-2*sd(X); U=mean(X)+2*sd(X)
Out=c() # storage for outlier
for(j in X){
  if(j<L|j>U){o=j}else{o=NULL}
  Out=c(Out,o) # append o to the vector of Out
}
Out
#there is no outliers
```

#Quiz:
consider **Seed_Data** , and **P** as the variable of interest.

a) Specify the outliers of **P** based on one-sigma interval estimation

b) Specify the outliers of **P** based on box-plot technique.


```python
%%R
data=read.csv('Seed_Data.csv')
X=data$P
#one-sigma interval estimation
L=mean(X)-1*sd(X); U=mean(X)+1*sd(X)
Out_z=c()  # outliers based on one-sigma
for (i in X){
    if (i<L || i> U){o=i}else{o=NULL}
    Out_z=c(Out_z,o)
}
print('---outliers for one-sigma interval estimation---')
print(Out_z)
##################boxplot technique#######################
Q=quantile(X)
Q1=Q[2]; Q3=Q[4]; IQR=IQR(X)
Lb=Q1-1.5*IQR ; Ub=Q3+1.5*IQR
Out_Box=c()
for (i in X){
    if (i<Lb || i> Ub){o=i}else{o=NULL}
    Out_Box=c(Out_Box,o)
}
print('---outliers based on box-plot technique---')
print(Out_Box)
```

## missing value imputation
In statistics, imputation is the process of replacing missing data with substituted values such as **mean**, **median**, **mode**, etc.

**Illustrative Example 16:**

Use **airquality** dataset, check whether is any missing values in this dataset. Remove missing values and print out the cleaned dataset.

```python
%%R
data=airquality
print(head(data))
mean(data$Ozone)
```

```python
#remove missing values
%%R
newdata = na.omit(data)
head(newdata)
```

**Illustrative Example 17:**
Impute the missing values with mean of **Ozone** in the **airquality** dataset.

```python
%%R
# Create mean
m=mean(data$Ozone,na.rm = TRUE)
print(m)
data$Ozone[is.na(data$Ozone)]= m
head(data,20)
```

#Illustrative Example 20:

 Impute the missing values with mode of **Ozone** in the **airquality** dataset.

```python
%%R
data2=airquality
X=data2$Ozone
y=table(X)
mode=names(y)[which(y==max(y))]

data2$Ozone[is.na(data2$Ozone)]= mode
print(mode)
head(data2)
```
