**Example 1:(Convert Celsius to Fahrenheit)**

```python
# Write a program that reads a Celsius degree from
# the console and converts it to Fahrenheit and displays the result. The formula for
# the conversion is as follows:
# fahrenheit = (9 / 5) * celsius + 32
#  Converting Celsius temperature to Fahrenheit
inp_temp = float(input("Enter the temperature in celsius : "))
out_temp = (inp_temp * (9/5)) + 32
print(f"The temperature of {inp_temp} celsius in Fahrenhti is {out_temp}")
```

**Example 2:(Compute the volume of a cylinder)**

```python
# Write a program that reads in the radius and
# length of a cylinder and computes the area and volume using the following formulas:
# area = radius * radius * π
# volume = area * length

PI = 3.1415
inp_len = float(input("Enter the lenght of Cylinder :" ))
inp_rad = float(input("Enter the radius of the Cylinder :  "))
area = inp_rad * inp_rad * PI
volume = inp_len * area
print(f"The area and volume for the Cylinder with lenght {inp_len} and raidus {inp_rad} is {area} and {volume}")
```

**Example 5:(Sum the digits in an integer)**

```python
# Write a program that reads an integer between 0 and
# 1000 and adds all the digits in the integer. For example, if an integer is 932, the
# sum of all its digits is 14. (Hint: Use the % operator to extract digits, and use the //
# operator to remove the extracted digit. For instance, 932 % 10 = 2 and 932 //
# 10 = 93.)
inp_int = int(input("Enter the Integer :" ))
inp_int_copy = inp_int
sum_digits = 0
while inp_int != 0:
    digit = inp_int % 10
    sum_digits = sum_digits + digit
    inp_int = inp_int // 10

print(f"Sum of digits {inp_int_copy} is {sum_digits}")
```

**Example 6:(Find the number of years and days)**

```python
# Write a program that prompts the user to
# enter the minutes (e.g., 1 billion), and displays the number of years and days for
# the minutes.
inp_min = int(input("Enter the Minutes :" ))
no_years = inp_min // 525600
no_days = (inp_min % 525600) // 1440
rem_min = (inp_min % 525600) % 1440
print(f"Years {no_years}, Days {no_days}, Minutes {rem_min}")
```

**Example 7: (Science: calculate energy)**

```python
#Write a program that calculates the energy needed to
# heat water from an initial temperature to a final temperature. Your program should
# prompt the user to enter the amount of water in kilograms and the initial and final
# temperatures of the water. The formula to compute the energy is
# Q = M * (finalTemperature – initialTemperature) * 4184
# where M is the weight of water in kilograms, temperatures are in degrees Celsius,
# and energy Q is measured in joules.
inp_water = float(input("Enter the Water in Kg:" ))
init_temp = float(input("Enter the Initial Temperature :" ))
fin_temp = float(input("Enter the Final Temperature :" ))
energy = (fin_temp - init_temp) * 4184 * inp_water
print("Energy Needed in Joules is ",energy)
```
