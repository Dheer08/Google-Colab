```python
data['A Stage']=data['A Stage'].map({'Regional':0, 'Distant':1})
```

```python
data['Estrogen Status']=data['Estrogen Status'].map({'Positive':0, 'Negative':1})
```

```python
data['Progesterone Status']=data['Progesterone Status'].map({'Positive':0, 'Negative':1})
```

```python
data['Status']=data['Status'].map({'Alive':0, 'Dead':1})
```

```python
data.info()
```

```python
features=['Race','Marital Status','T Stage ','N Stage','6th Stage','differentiate','Grade']
data2=get_dummies(data,columns=features)
```

```python
data2.info()
```

```python
Y=data2['Status'] # Dependent variable
X=data2.drop('Status',axis=1) # Axis=1 drop column, Independent columns/ dataset
```

```python
X_scaled= StandardScaler().fit_transform(X)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size = 0.3, random_state = 100)
```
