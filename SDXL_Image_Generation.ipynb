```python
# --- Install dependencies ---
!pip install diffusers transformers accelerate safetensors torch torchvision

# --- Import libraries ---
import torch
from diffusers import StableDiffusionXLPipeline
from pathlib import Path
from PIL import Image
import random

# --- Choose device ---
device = "cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu"
print(f"Running on device: {device}")

# --- Load SDXL model ---
model_id = "stabilityai/stable-diffusion-xl-base-1.0"
pipe = StableDiffusionXLPipeline.from_pretrained(model_id, torch_dtype=torch.float16 if device=="cuda" else torch.float32)
pipe = pipe.to(device)

# --- Optional: enable memory optimization ---
pipe.enable_attention_slicing()

# --- Output directory ---
output_dir = Path("data/sdxl_faces")
output_dir.mkdir(parents=True, exist_ok=True)

# --- Prompts to diversify the dataset (can be expanded later) ---
prompts = [
    "portrait photo of a young woman, smiling, natural lighting, high resolution",
    "portrait photo of an elderly man with glasses, studio lighting, realistic skin texture",
    "close-up photo of a middle-aged man, confident expression, professional background",
    "portrait of a teenage girl with curly hair, outdoor lighting, realistic tone",
    "photo of a woman with dark hair and brown eyes, cinematic lighting",
    "photo of a man with a beard, soft background bokeh, professional headshot"
]

# --- Number of images to generate ---
num_images = 50  # Adjust as needed, e.g., 1000 for final dataset

# --- Generation loop ---
for i in range(num_images):
    prompt = random.choice(prompts)
    image = pipe(prompt=prompt, num_inference_steps=30, guidance_scale=7.5).images[0]

    # Resize to 512x512 and save
    image = image.resize((512, 512), Image.LANCZOS)
    image.save(output_dir / f"synthetic_face_{i+1:04d}.png")

    if (i + 1) % 10 == 0:
        print(f"Generated {i + 1} / {num_images} images")

print("✅ Dataset generation complete!")
print(f"Saved {num_images} images to: {output_dir.resolve()}")
```
